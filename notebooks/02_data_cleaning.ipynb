### Findings — Missing Order Approval Timestamps

- A total of 160 orders have missing `order_approved_at` values.
- Of these, 141 orders are cancelled, 14 are marked delivered, and 5 are marked created.
- All 14 delivered orders have recorded carrier handover and customer delivery timestamps despite missing approval timestamps.
- These inconsistencies may reflect incomplete timestamp records, but their exact cause is unknown.
- All affected records will be retained, and missing approval timestamps will not be imputed.

### Findings — Missing Carrier Handover Timestamps

- A total of 1,783 orders have missing carrier handover timestamps.
- Most belong to orders with statuses other than `delivered`.
- Two orders are marked `delivered` despite missing carrier handover timestamps.
- One of these orders has a recorded customer delivery date, while the other also has missing approval and customer delivery timestamps.
- All affected records will be retained, and missing carrier handover timestamps will not be imputed.

```python
carrier_before_approval = (
    orders['order_delivered_carrier_date']
    < orders['order_approved_at']
)

customer_before_carrier = (
    orders['order_delivered_customer_date']
    < orders['order_delivered_carrier_date']
)

print("Carrier before approval:", carrier_before_approval.sum())
print("Customer before carrier:", customer_before_carrier.sum())
```

```python
customer_before_purchase = (
    orders['order_delivered_customer_date']
    < orders['order_purchase_timestamp']
)

print(
    "Customer delivery before purchase:",
    customer_before_purchase.sum()
)
```

### Findings — Timestamp Validation

- 1,359 records have carrier handover timestamps earlier than their approval timestamps.
- 23 records have customer delivery timestamps earlier than their carrier handover timestamps.
- No records have customer delivery timestamps earlier than their purchase timestamps.
- The identified chronological inconsistencies will be retained for further investigation. No timestamps have been modified or removed.
